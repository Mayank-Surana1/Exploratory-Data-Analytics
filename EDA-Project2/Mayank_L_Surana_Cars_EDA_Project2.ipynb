##Now Eliminating the Duplicate Rows

```python
print('Before removing the Duplicates',df.shape)
df = df.drop_duplicates()
print('After removing the Duplicates',df.shape)  #We can Clearly see that - 5 rows are removed whereas there are No duplicate Columns
```

##Handle missing values

```python
df.isnull().sum()  #Clearly we can see that ['fueltype','enginesize','horsepower','price']  has the null Values
```

```python
#Handle missing values in numerical columns
numerical_missing = ["horsepower", "price", "enginesize"]

for column in numerical_missing:
    df.loc[:, column] = df[column].fillna(df[column].median())


#Handle missing value in categorical column
df.loc[:, "fueltype"] = df["fueltype"].fillna(df["fueltype"].mode()[0])
```

```python
rows_after = df.shape[0]

print("\nDATA CLEANING - AFTER")
print("Rows:", rows_after)
print("Columns:", df.shape[1])
print("Missing cells:", df.isnull().sum().sum())  #By now we have removed all the missing cells
print("Duplicate rows:", df.duplicated().sum())   #By no we have removed all the duplicated rows
```

##Univariate Analysis

###Histogram Plot

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x="price", bins=20, kde=True)  #to create a histogram.  #Bins divide range of prices into approximately 20 intervals/groups.

plt.title("Distribution of Car Prices")
plt.xlabel("Price")
plt.ylabel("Number of Cars")

plt.show()
```

###Count Plot

```python
plt.figure(figsize=(4, 4))

sns.countplot(data=df, x="fueltype")

plt.title("Distribution of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")

plt.show()
```

###Histogram Plot

```python
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="horsepower", bins=20, kde=True)

plt.title("Distribution of Car Horsepower")
plt.xlabel("Horsepower")
plt.ylabel("Number of Cars")

plt.show()
```

##Bivariate Analysis

###Scatter Plot

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,x="horsepower",y="price",color="green")

plt.title("Car Price vs Horsepower")
plt.xlabel("Horsepower")
plt.ylabel("Price")

plt.show()
```

###Scatter Plot

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x="enginesize",y="price",color="red")
plt.title("Car Price vs Engine Size")
plt.xlabel("Engine Size")
plt.ylabel("Price")
plt.show()
```

###Box Plot

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df,x="fueltype",y="price")

plt.title("Car Price Distribution by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Price")

plt.show()
```

##Multivariate Analysis

```python
plt.figure(figsize=(10,7))
correlation = df[numerical_columns].corr()
sns.heatmap(
    correlation,   ##For Multiple variables we Use Corelation Heatmap
    annot=True,cmap="coolwarm",fmt=".1f")

plt.title("Correlation Heatmap of Car Features")
plt.show()
```

##Outlier Analysis

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="price",
    color="steelblue"
)

plt.title("Box Plot of Car Prices")
plt.xlabel("Price")

plt.show()
```
